```python
# Loading summaries and combining into a single plot

summary_paths = [
    r"C:\Users\chloe\DissertationSimulations\Result summaries\1D electrodes\exp1_st1_results1_ts_exp_matrix_cols_0-1.json",
    r"C:\Users\chloe\DissertationSimulations\Result summaries\1D electrodes\exp1_st1_results2_psd_exp_matrix_cols_0-1.json",
    r"C:\Users\chloe\DissertationSimulations\Result summaries\1D electrodes\exp1_st1_results3_specmod_exp_matrix_cols_0-1.json"
]

labels = [
    "Average time series",
    "Average PSD",
    "Average spectral model"
]

fig, ax = plot_summary_boxplots(
    summary_paths=summary_paths,
    labels=labels,
    true_mean_exp=np.mean(-exp_matrix)
)

fig.savefig(r"C:\Users\chloe\DissertationSimulations\Figures\1D electrodes\Combined plots\exp1_st1_combined_plot_1D_exp_matrix_cols_0-1.png", dpi=300, bbox_inches="tight")
```
