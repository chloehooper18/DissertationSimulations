### Simulating 2D lines of electrodes 

- Simulating electrode clusters where the exponent is a random value between -1 and -2
- Average the PSD and plot the Spectral Model
- Simulate this for aperiodic signals only

##### Imports

```python
# Imports required for functionality 
import numpy as np
import matplotlib.pyplot as plt

from dissertation_simulations.params import cz_sim_params_ap

from dissertation_simulations.generate_electrodes import generate_2D_electrodes_ap
from dissertation_simulations.psd import avg_psd_2D
from dissertation_simulations.exponents import gen_exponents
from dissertation_simulations.params import spectral_model_params

from specparam import SpectralModel
from dissertation_simulations.params import freq_range
```

#### 2x2 electrode cluster
- Simulate a 2x2 electrode with random exponents between -1 and -2
- Calculate the average PSD and plot the spectral model

```python
# Load in copy of parameters
params = cz_sim_params_ap.copy()

# Generate random exponents
exponents = gen_exponents(4, (-2, -1))
print(exponents)

# Generate electrode cluster
grid_signals, times = generate_2D_electrodes_ap(2, 2, exponents)

# Compute and plot average PSD
psds, avg_powers, freqs = avg_psd_2D(grid_signals, params)

# Fit and plot Spectral model
fm = SpectralModel(**spectral_model_params)
freq_range = freq_range

fm.report(freqs, avg_powers, freq_range)
```
