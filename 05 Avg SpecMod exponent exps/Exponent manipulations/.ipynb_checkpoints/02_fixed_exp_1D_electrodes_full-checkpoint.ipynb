### Averaging Spectral Models for multiple ppts
- Function that calculates average Spectral Model exponent for n repeats
- Function uses base params but these can be overridden to allow for different manipulations
- Notebook contains full signals only

#### Experiment 1
- Simulate 3 electrodes where the exponent is -1 for each, all other params use the pre-set base params, full signals
- Calculate PSD for each
- Average the Spectral Model exponent
- Repeat this simulation n times
- Plot all exponent outputs and look at the distribution - retrieve mean and SD values

```python
from dissertation_simulations.spec_mod import avg_specmod_exponent_distribution, sweep_electrodes_specmod_1D
from dissertation_simulations.params import electrode_sim_params_full
```

```python
avg_specmod_exponent_distribution(n_repeats=100, n_electrodes=3, base_params=electrode_sim_params_full, exponent=-1)
```
