- Function uses base params but these can be overridden to allow for different manipulations
- Notebook contains full signals only

- Calculate PSD for each
- Average the Spectral Model exponent
- Repeat this simulation n times
- Plot all exponent outputs and look at the distribution - retrieve mean and SD values

```python
# Imports required for functionality 
from dissertation_simulations.spec_mod import avg_specmod_exponent_distribution_2D, sweep_electrodes_specmod_2D
from dissertation_simulations.params import electrode_sim_params_full
```
