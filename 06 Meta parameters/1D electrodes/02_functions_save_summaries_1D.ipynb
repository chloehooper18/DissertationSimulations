### Running sweep functions and saving the result summaries

- call functions
- do not produce plot
- save result summaries

```python
# Imports for functionality
from dissertation_simulations.time_series import sweep_electrodes_ts_1D, sweep_electrodes_ts_2D
from dissertation_simulations.psd import sweep_electrodes_psd_1D, sweep_electrodes_psd_2D
from dissertation_simulations.spec_mod import sweep_electrodes_specmod_1D, sweep_electrodes_specmod_2D
from dissertation_simulations.exponents import generate_spatial_exponent_matrix

import json
import numpy as np
```

#### Experiment 1

```python
electrodes, means, stds, errors, fig = sweep_electrodes_ts_1D(electrode_range=range(1, 10), n_repeats=100, exponent=-1.5, plot=False)

exp1_result1_ts = {
    "electrodes": electrodes.tolist(),
    "means": means.tolist(),
    "stds": stds.tolist(),
    "errors": errors.tolist(),
    "exponent": -1.5,
    "n_repeats": 100
}

with open(r"C:\Users\chloe\DissertationSimulations\Result summaries\1D electrodes\exp1results1_ts_exponent1.5.json", "w") as f:
    json.dump(exp1_result1_ts, f, indent=4)
```

```python
grid_range = [(1, n) for n in range(1, 10)]

exponent_matrices = [generate_spatial_exponent_matrix(grid_shape=grid_shape, exp_min=-2, exp_max=-0.5, pattern="radial")
    for grid_shape in grid_range]

electrodes, means, stds, errors, fig = sweep_electrodes_ts_2D(grid_range=grid_range, exponent_matrices=exponent_matrices, 
                                                              n_repeats=100, plot=False)

exp2_result1_ts = {
    "electrodes": electrodes.tolist(),
    "means": means.tolist(),
    "stds": stds.tolist(),
    "errors": errors.tolist(),
    "n_repeats": 100
}

with open(r"C:\Users\chloe\DissertationSimulations\Result summaries\1D electrodes\exp2results1_ts_exp_matrix_radial_-0.5-2.json", "w") as f:
    json.dump(exp2_result1_ts, f, indent=4)
```
