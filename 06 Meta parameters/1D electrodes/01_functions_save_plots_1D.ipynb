```python
grid_range = [(1, n) for n in range(1, 10)]

exponent_matrices = [generate_spatial_exponent_matrix(
        grid_shape=grid_shape,
        exp_min=-2,
        exp_max=-0.5,
        pattern="radial")

    for grid_shape in grid_range]

electrodes, means, stds, errors, fig = sweep_electrodes_ts_2D(grid_range=grid_range, exponent_matrices=exponent_matrices, 
                                                              n_repeats=100)
fig.savefig(r"C:\Users\chloe\DissertationSimulations\Figures\1D electrodes\exp2fig1_ts_exponent_radial0.5-2.png", dpi=300)
```
