### Running sweep functions and saving the result summaries

- call functions
- do not produce plot
- save result summaries

```python
# Imports for functionality
from dissertation_simulations.time_series import sweep_electrodes_ts_2D
from dissertation_simulations.psd import sweep_electrodes_psd_2D
from dissertation_simulations.spec_mod import sweep_electrodes_specmod_2D
from dissertation_simulations.exponents import generate_2D_exponent_matrix

import json
import numpy as np
```

#### Experiment 1

```python
grid_range = [(n, n) for n in range(2, 9)]

exponent_matrices = [generate_2D_exponent_matrix
                     (grid_shape=grid_shape, exp_range=(-2, -1), gradient_axis="cols")
    for grid_shape in grid_range]

electrodes, means, stds, errors, fig = sweep_electrodes_ts_2D(grid_range=grid_range, exponent_matrices=exponent_matrices, 
                                                              n_repeats=100, plot=False)

exp1_result1_ts = {
    "electrodes": electrodes.tolist(),
    "means": means.tolist(),
    "stds": stds.tolist(),
    "errors": errors.tolist(),
    "exponent": -1.5,
    "n_repeats": 100
}

with open(r"C:\Users\chloe\DissertationSimulations\Result summaries\2D electrodes\exp1results1_ts_exp_matrix_cols_-1-2.json", "w") as f:
    json.dump(exp1_result1_ts, f, indent=4)
```
